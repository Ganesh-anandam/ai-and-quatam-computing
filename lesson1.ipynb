```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer

qc=QuantumCircuit(1,1)
qc.h(0)

qc.measure(0,0)

simulotor=Aer.get_backend('aer_simulator')
job=simulotor.run(qc)
result=job.result()
counts=result.get_counts()
print(counts)
```
